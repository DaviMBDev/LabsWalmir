### 3. Verificação das Bibliotecas Essenciais
Validamos se os pacotes fundamentais (`pandas`, `matplotlib`, `seaborn`, `sklearn`) estão devidamente instalados e acessíveis:

````python
pacotes = ['pandas', 'matplotlib', 'seaborn', 'sklearn']
status = {}

for p in pacotes:
    try:
        mod = __import__(p)
        ver = getattr(mod, '__version__', 'instalado')
        status[p] = f"OK (v{ver})"
    except ImportError as e:
        status[p] = f"FALHA: {e}"

import pandas as pd
df_status = pd.DataFrame(list(status.items()), columns=['Biblioteca', 'Status'])
df_status
````

### 4. Teste Prático com Visualização (Seaborn Penguins)
Para testar a capacidade de renderização gráfica e bibliotecas científicas, carregamos o dataset clássico dos pinguins da Ilha Palmer e geramos um gráfico conjunto de dispersão com distribuições marginais.

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração de estilo visual
sns.set_theme(style="whitegrid", palette="muted")

# Carregamento do dataset oficial do Seaborn
df_penguins = sns.load_dataset("penguins")
print(f"Dataset carregado com {df_penguins.shape[0]} registros.")
df_penguins.head()
````

````python
# Informações estruturais e estatísticas rápidas
print(df_penguins.info())
df_penguins.describe()
````

````python
# Teste visual solicitado no Laboratório: Jointplot
# flliper_length_mm (comprimento da nadadeira) x bill_length_mm (comprimento do bico) por espécie
g = sns.jointplot(
    data=df_penguins,
    x="flipper_length_mm",
    y="bill_length_mm",
    hue="species",
    height=7,
    marginal_kws=dict(bins=15, fill=True)
)

g.fig.suptitle("Dispersão e Densidade Marginal: Nadadeira vs. Bico por Espécie", y=1.02, fontsize=14)
plt.show()
````
