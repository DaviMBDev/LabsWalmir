### Parte 1 — Primeiros Passos e Criação de Dados
Para utilizar o Pandas, importamos a biblioteca sob o alias convencional `pd`.

```python
import pandas as pd
import numpy as np

print(f"Versão do Pandas: {pd.__version__}")
```

#### 1.1 DataFrame
Um **DataFrame** é uma tabela bidimensional com linhas e colunas rotuladas. Cada coluna pode ter um tipo de dado diferente.

```python
# Criando um DataFrame a partir de um dicionário
df_yes_no = pd.DataFrame({'Yes': [50, 21], 'No': [131, 2]})
df_yes_no
```

As entradas de um DataFrame não estão limitadas a números inteiros — podem conter strings, floats e booleanos:

```python
df_reviews_simples = pd.DataFrame({
    'Bob': ['I liked it.', 'It was awful.'],
    'Sue': ['Pretty good.', 'Bland.']
})
df_reviews_simples
```

Por padrão, o índice das linhas é uma sequência numérica iniciando em 0. Podemos customizar esse índice usando o parâmetro `index`:

```python
df_produtos = pd.DataFrame({
    'Bob': ['I liked it.', 'It was awful.'],
    'Sue': ['Pretty good.', 'Bland.']
}, index=['Product A', 'Product B'])

df_produtos
```

#### 1.2 Series
Uma **Series** é uma sequência unidimensional de dados. Se o DataFrame é uma tabela inteira, a Series é uma única coluna isolada.

```python
# Series simples
serie_numeros = pd.Series([1, 2, 3, 4, 5])
serie_numeros
```

```python
# Series com índice personalizado e nome
serie_vendas = pd.Series(
    [30, 35, 40],
    index=['2015 Sales', '2016 Sales', '2017 Sales'],
    name='Product A'
)
serie_vendas
```

#### 1.3 Leitura e Escrita de Arquivos (CSV)
Na maior parte dos projetos, trabalhamos com dados existentes em arquivos como CSV (*Comma-Separated Values*).
Podemos ler com `pd.read_csv()` e salvar com `.to_csv()`.

```python
# Exemplo: criando dados e salvando em CSV para demonstração
df_demo = pd.DataFrame({
    'id': [1, 2, 3],
    'nome': ['Vinho Tinto', 'Vinho Branco', 'Espumante'],
    'pontos': [88, 92, 95]
})
df_demo.to_csv("dados_demo.csv", index=False)
print("Arquivo dados_demo.csv gravado com sucesso.")

# Lendo o CSV de volta
df_lido = pd.read_csv("dados_demo.csv")
print("Shape:", df_lido.shape)
df_lido.head()
```

### Parte 2 — Indexing, Selecting and Assigning (loc e iloc)

Selecionar pontos de dados específicos de maneira rápida e correta é um dos fundamentos da análise de dados.

#### 2.1 Acessores Nativos
Podemos acessar colunas como atributos (`df.coluna`) ou por chave (`df['coluna']`). O uso de colchetes é mais seguro, pois aceita espaços e caracteres reservados.

```python
# Criando dataset de exemplo com colunas do domínio
df_carros = pd.read_csv("CarPrice_Assignment.csv") if os.path.exists("CarPrice_Assignment.csv") else pd.DataFrame({
    'CarName': ['alfa-romero giulia', 'audi 100 ls', 'bmw x1'],
    'price': [13495, 13950, 16430],
    'fueltype': ['gas', 'gas', 'gas']
})

# Acesso por atributo
print("Primeiros carros (atributo):")
print(df_carros.CarName.head(3))

# Acesso por colchetes
print("\nPrimeiros carros (colchetes):")
print(df_carros['CarName'].head(3))

# Valor específico
print("\nPrimeiro elemento:", df_carros['CarName'][0])
```

#### 2.2 Seleção Posicional: `iloc`
O `iloc` seleciona linhas e colunas com base em suas **posições numéricas** inteiras (row-first, column-second).
- **Atenção:** Segue as regras do Python nativo: fatias do tipo `0:3` pegam os índices `0, 1, 2` (o fim é **excludente**).

```python
print("1. Primeira linha inteira:")
print(df_carros.iloc[0])

print("\n2. Primeira coluna inteira (primeiras 5 linhas):")
print(df_carros.iloc[:, 0].head(5))

print("\n3. Primeiras 3 linhas da primeira coluna:")
print(df_carros.iloc[:3, 0])

print("\n4. Linhas de índice 1 a 2 da primeira coluna:")
print(df_carros.iloc[1:3, 0])

print("\n5. Lista de posições específicas [0, 1, 2]:")
print(df_carros.iloc[[0, 1, 2], 0])

print("\n6. Últimas 3 linhas do DataFrame (índice negativo):")
print(df_carros.iloc[-3:])
```

#### 2.3 Seleção por Rótulo: `loc`
O `loc` seleciona dados com base no **rótulo/nome** do índice e da coluna.
- ⚠️ **Gotcha importante:** `loc` indexa de forma **inclusiva**: `0:3` seleciona as linhas `0, 1, 2, 3`!

```python
# Selecionando pelo índice da linha e nome da coluna
print("1. Linha 0, coluna 'CarName':", df_carros.loc[0, 'CarName'] if 'CarName' in df_carros.columns else "N/A")

# Selecionando múltiplas colunas
colunas_interesse = [c for c in ['CarName', 'carbody', 'price'] if c in df_carros.columns]
print("\n2. Linhas 0 a 3 das colunas de interesse:")
df_carros.loc[0:3, colunas_interesse]
```

### Resumo: Regra Prática
| Acessor | Critério | Fatiamento `0:2` | Uso Recomendado |
| :--- | :--- | :--- | :--- |
| **`iloc`** | Posição numérica (0, 1, 2...) | `[0, 1]` (exclusivo) | Tratar o dataset como matriz por coordenadas numéricas |
| **`loc`** | Rótulo / Nome (label) | `[0, 1, 2]` (inclusivo) | Seleção por nome de coluna ou índice com significado |
